# Sensitivity Analysis: Prior Utilization and Follow-Up Timing

This notebook runs the reproducible sensitivity-analysis workflow for the synthetic EHR readmission project. It focuses on prior utilization stratification, observed outpatient follow-up timing windows, and adjusted model comparisons with and without follow-up covariates.

Outputs are aggregate tables only. No patient-level records are exported.

## Why This Sensitivity Pass Matters

The primary dashboard KPIs show cohort size, readmission rate, follow-up, and ED revisit rates. This sensitivity pass adds the analytic story behind those KPIs by asking whether readmission patterns differ by prior utilization burden and whether adjusted model results change when observed follow-up timing variables are included.

Outpatient follow-up is treated as an observed utilization measure, not a causal exposure. Patients may be readmitted before follow-up can occur, and low event counts can make follow-up model terms unstable.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from scripts.run_sensitivity_analysis import run_sensitivity_analysis

outputs = run_sensitivity_analysis()
```
